```python
import pandas as pd

df = pd.read_csv(
    r'F:\Downloads\Sales-Analytics-Project\Data\Row\Global_Superstore2.csv',
    encoding='latin1'
)

df.head()
```

```python
print(df['Row ID'].nunique())
print(df.shape[0])
df = df.drop(columns=['Row ID'])
```

```python
df['Customer ID'] = df['Customer ID'].astype(str)
df['Product ID'] = df['Product ID'].astype(str)
df['Order ID'] = df['Order ID'].astype(str)
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')
```

```python
df[['Order Date', 'Ship Date']].isnull().sum()
```

```python
text_columns = [
    'Order ID',
    'Customer Name',
    'Segment',
    'City',
    'State',
    'Country',
    'Region',
    'Category',
    'Sub-Category',
    'Product Name',
    'Ship Mode',
    'Order Priority'
]
for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df['Postal Code'].isnull().sum()
df = df.drop(columns=['Postal Code'])
```

```python
print("Negative Sales:", (df['Sales'] < 0).sum())
print("Invalid Quantity:", (df['Quantity'] <= 0).sum())
print("Invalid Discount:", ((df['Discount'] < 0) | (df['Discount'] > 1)).sum())
print("Negative Shipping Cost:", (df['Shipping Cost'] < 0).sum())
```

```python
(df['Ship Date'] < df['Order Date']).sum()
```
