Numeric Validation

```python
print("Invalid Sales:", (df['Sales'] < 0).sum())
print("Invalid Quantity:", (df['Quantity'] <= 0).sum())
print("Invalid Discount:", (df['Discount'] < 0).sum())
print("Invalid Discount > 1:", (df['Discount'] > 1).sum())
print("Invalid Shipping Cost:", (df['Shipping Cost'] < 0).sum())
```

```python
print(df['Ship Mode'].unique())
print(df['Segment'].unique())
print(df['Category'].unique())
print(df['Sub-Category'].unique())
print(df['Order Priority'].unique())
print(df['Region'].unique())
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')
print("Invalid Ship Dates:", (df['Ship Date'] < df['Order Date']).sum())
print("Invalid Order Dates:", df['Order Date'].isna().sum())
print("Invalid Ship Dates:", df['Ship Date'].isna().sum())
```

```python
for col in ['Ship Mode', 'Segment', 'Category', 'Sub-Category', 'Order Priority', 'Region']:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))
```

```python
print("Zero Sales:", (df['Sales'] == 0).sum())
print("Zero Quantity:", (df['Quantity'] == 0).sum())
print("Zero Shipping Cost:", (df['Shipping Cost'] == 0).sum())
```
